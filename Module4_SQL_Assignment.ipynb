## Exercise 16.4 - Creating a SQLite Database

```python
import sqlite3

# Connect to the database
conn = sqlite3.connect('books.db')

# Create a cursor
cursor = conn.cursor()

# Create the books table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS books (
        title TEXT,
        author TEXT,
        year INTEGER
    )
''')

#Save changes
conn.commit()

print("Database and books table created successfully.")
```

## Exercise 16.5 - Reading CSV Data into SQLite

```python

import csv

# Open the CSV file and read its contents
with open('books2.csv', 'r', newline='', encoding='utf-8') as file:
    reader = csv.DictReader(file)

    for row in reader:
        cursor.execute(
            'INSERT INTO books (title, author, year) VALUES (?, ?, ?)',
            (row['title'], row['author'], int(row['year']))
        )

conn.commit()

print("Five books successfully added to the books table.")
```

## Exercise 16.8 - Connecting to SQLite Using SQLAlchemy

```python
from sqlalchemy import create_engine, text

# Connect to the SQLite database
engine = create_engine('sqlite:///books.db')

# Retrieve book titles in alphabetical order
with engine.connect() as connection:
    result = connection.execute(
        text('SELECT title FROM books ORDER BY title')
    )

    for row in result:
        print(row[0])
```
