```python
# Exercise 4.2

small = True
green = True

if small:
    if green:
        print("pea")
    else:
        print("cherry")
else:
    if green:
        print("watermelon")
    else:
        print("pumpkin")
```

```python
# Exercise 6.2

guess_me = 7
number = 1

while number <= guess_me:
    if number < guess_me:
        print('too low')
    elif number == guess_me:
        print('found it!')
        break
    else:
        print('oops')
        break

    number += 1
```

```python
# Exercise 6.3

guess_me = 5

for number in range(10):
    if number < guess_me:
        print('too low')
    elif number == guess_me:
        print('found it!')
        break
    else:
        print('oops')
        break

    number += 1
```
